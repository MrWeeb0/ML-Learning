## Exercise 2: Merge All CSVs

**Task**: Write a function `merge_all()` that:

1. Loads `students.csv`, `courses.csv`, and `grades.csv`.
2. Merges them into a single DataFrame using `student_id` |and `course_id`.
3. Returns the merged DataFrame.

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>pd.merge</code> twice: first on <code>student_id</code>, then on <code>course_id</code>.
</details>

```python
def merge_all():
    students = pd.read_csv("students.csv")
    courses = pd.read_csv("courses.csv")
    grades = pd.read_csv("grades.csv")
    
    merged = pd.merge(grades, students, on="student_id")
    merged = pd.merge(merged, courses, on="course_id")
    
    return merged
```

```python
# Tests
df = merge_all()
assert "student_id" in df.columns
assert "grade" in df.columns
assert "course_name" in df.columns
assert df.shape[0] >= 30
```

## Exercise 3: Average Grade per Student

**Task**: Write a function `average_grade_per_student()` that returns a DataFrame with one row per student and their average grade across all courses.

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>groupby</code> on <code>student_name</code> and apply <code>mean()</code> on <code>grade</code>.
</details>

```python
def average_grade_per_student():
    df = merge_all()
    result = df.groupby("student_name")["grade"].mean().reset_index()
    result.columns = ["student_name", "avg_grade"]
    return result
```

```python
# Tests
df = average_grade_per_student()
assert "student_name" in df.columns
assert "avg_grade" in df.columns
assert df.shape[0] == 10
```

## Exercise 6: Gender Distribution by Course

**Task**: Write a function `gender_distribution()` that returns a table (DataFrame) showing how many male and female students are enrolled in each course.

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>groupby</code> with <code>size()</code> and <code>unstack()</code> to pivot the gender counts.
</details>

```python
def gender_distribution():
    df = merge_all()
    result = df.groupby(["course_name", "gender"]).size().unstack(fill_value=0)
    return result
```

```python
# Tests
df = gender_distribution()
assert "F" in df.columns and "M" in df.columns
assert df.shape[0] == 6
```

## Exercise 10: Normalize the Grade Column

**Task**: Write a function `normalize_grades()` that returns a DataFrame with a new column `grade_norm` where each grade is normalized using z-score:  
`z_score = (grade - mean) / std`

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Compute mean and std of <code>grade</code>, then use <code>apply(lambda x: (x - mean) / std)</code>.
</details>

```python
def normalize_grades():
    df = merge_all()
    mean = df["grade"].mean()
    std = df["grade"].std()
    df["grade_norm"] = (df["grade"] - mean) / std
    return df
```

```python
# Tests
df = normalize_grades()
assert "grade_norm" in df.columns
assert abs(df["grade_norm"].mean()) < 1e-6  # close to 0
```

## Optional Exercise 1: Courses Never Taken by Any Female Student

## Optional Exercise 3: Grade Progress Per Student (Cumulative Mean)

**Task**: Write a function `grade_progress_per_student()` that returns a DataFrame showing how each student's average grade evolves over time, ordered by `course_name` alphabetically.

- Include columns: `student_name`, `course_name`, `grade`, `avg_so_far`
- The `avg_so_far` column should contain the cumulative average grade for each student, computed as they progress through the courses alphabetically.

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Sort the merged DataFrame by <code>student_name</code> and <code>course_name</code>, then use <code>groupby()</code> + <code>expanding().mean()</code>.
</details>

```python
def grade_progress_per_student():
    df = merge_all()
    df = df.sort_values(["student_name", "course_name"])
    df["avg_so_far"] = df.groupby("student_name")["grade"].expanding().mean().reset_index(level=0, drop=True)
    return df[["student_name", "course_name", "grade", "avg_so_far"]]
```

```python
# Tests
df = grade_progress_per_student()
assert all(col in df.columns for col in ["student_name", "course_name", "grade", "avg_so_far"])
assert df["avg_so_far"].notnull().all()
```
