## Exercise 3: Add a Read Tracker

**Task**: Update the `Book` class so that:

1. It has an additional attribute `pages_read`, initialized to 0.
2. Add a method `read(pages)` that adds the number of pages read to `pages_read`, but **never allows it to exceed the total number of pages**.
3. Add a method `progress()` that returns the percentage of the book read as an integer.

Example:
```python
b = Book("Short Book", "Me", 100)
b.read(30)
b.progress() → 30
b.read(100)
b.progress() → 100

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>min()</code> inside <code>read()</code> to ensure <code>pages_read</code> never exceeds <code>pages</code>. In <code>progress()</code>, use integer division or <code>int()</code> with the formula: <code>(pages_read / pages) * 100</code>.
</details>

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        self.pages_read = 0

    def summary(self):
        return f"Title: {self.title} | Author: {self.author} | Pages: {self.pages}"

    def read(self, pages):
        self.pages_read = min(self.pages_read + pages, self.pages)

    def progress(self):
        return int((self.pages_read / self.pages) * 100)
````

````python
# Tests
b = Book("Read Me", "Anon", 200)
b.read(50)
assert b.pages_read == 50
assert b.progress() == 25
b.read(300)
assert b.pages_read == 200
assert b.progress() == 100
````

## Exercise 4: Create a `Library` Class with a Collection

**Task**: Create a class `Library` that:

1. Stores a list of `Book` objects.
2. Has a method `add_book(book)` to add a book to the library.
3. Has a method `total_pages()` that returns the sum of pages of all books in the library.

You will need to reuse the `Book` class from previous exercises.

Example:
```python
lib = Library()
lib.add_book(Book("Book A", "A", 100))
lib.add_book(Book("Book B", "B", 200))
lib.total_pages() → 300

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Keep a list in the constructor, like <code>self.books = []</code>, and use <code>append()</code> to add books. Use <code>sum()</code> with a generator expression in <code>total_pages()</code>.
</details>

````python
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def total_pages(self):
        return sum(book.pages for book in self.books)
````

````python
# Tests
lib = Library()
lib.add_book(Book("A", "Author A", 100))
lib.add_book(Book("B", "Author B", 200))
assert lib.total_pages() == 300
````

## Exercise 5: Prevent Negative Page Counts

**Task**: Update the `Book` class to ensure:

- If someone tries to create a `Book` with a negative number of pages, raise a `ValueError` with the message:  
  `"Pages must be a non-negative integer."`

Example:
```python
Book("Broken Book", "Nobody", -5) → raises ValueError

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  In the <code>__init__</code> method, check if <code>pages</code> is less than 0 and raise an exception if so.
</details>

````python
class Book:
    def __init__(self, title, author, pages):
        if pages < 0:
            raise ValueError("Pages must be a non-negative integer.")
        self.title = title
        self.author = author
        self.pages = pages
        self.pages_read = 0

    def summary(self):
        return f"Title: {self.title} | Author: {self.author} | Pages: {self.pages}"

    def read(self, pages):
        self.pages_read = min(self.pages_read + pages, self.pages)

    def progress(self):
        return int((self.pages_read / self.pages) * 100)
````

````python
# Tests
try:
    Book("Bad", "None", -10)
    assert False, "Expected ValueError for negative pages"
except ValueError as e:
    assert str(e) == "Pages must be a non-negative integer."
````

## Exercise 8: Remove Book by Title

**Task**: Add a method `remove_book(title)` to the `Library` class that:

- Removes the first book found with the given title.
- If no book with that title exists, raise a `ValueError` with the message: `"Book not found."`

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Loop through the list of books and use <code>.remove()</code> to remove the matching one.
</details>

````python
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def total_pages(self):
        return sum(book.pages for book in self.books)

    def find_books_by_author(self, author):
        return [book for book in self.books if book.author == author]

    def remove_book(self, title):
        for book in self.books:
            if book.title == title:
                self.books.remove(book)
                return
        raise ValueError("Book not found.")
````

## Exercise 9: Sort Books by Page Count

**Task**: Add a method `sort_by_pages(descending=False)` to the `Library` class that:

- Returns a new list of books sorted by number of pages.
- If `descending=True`, the list should be sorted in descending order.

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>sorted()</code> with a lambda function as the <code>key</code>.
</details>

````python
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def total_pages(self):
        return sum(book.pages for book in self.books)

    def find_books_by_author(self, author):
        return [book for book in self.books if book.author == author]

    def remove_book(self, title):
        for book in self.books:
            if book.title == title:
                self.books.remove(book)
                return
        raise ValueError("Book not found.")

    def sort_by_pages(self, descending=False):
        return sorted(self.books, key=lambda book: book.pages, reverse=descending)
````

````python
# Tests
lib = Library()
lib.add_book(Book("A", "Auth", 50))
lib.add_book(Book("B", "Auth", 10))
lib.add_book(Book("C", "Auth", 30))

asc = lib.sort_by_pages()
desc = lib.sort_by_pages(descending=True)

assert [b.pages for b in asc] == [10, 30, 50]
assert [b.pages for b in desc] == [50, 30, 10]
````

## Exercise 10: Count Books by Title Keyword

**Task**: Add a method `count_books_with(keyword)` to the `Library` class that:

- Returns the number of books that contain the given keyword (case-insensitive) in their title.

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>keyword.lower() in book.title.lower()</code> to check case-insensitively.
</details>

````python
class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def total_pages(self):
        return sum(book.pages for book in self.books)

    def find_books_by_author(self, author):
        return [book for book in self.books if book.author == author]

    def remove_book(self, title):
        for book in self.books:
            if book.title == title:
                self.books.remove(book)
                return
        raise ValueError("Book not found.")

    def sort_by_pages(self, descending=False):
        return sorted(self.books, key=lambda book: book.pages, reverse=descending)

    def count_books_with(self, keyword):
        return sum(1 for book in self.books if keyword.lower() in book.title.lower())
````

````python
# Tests
lib = Library()
lib.add_book(Book("Python Basics", "A", 100))
lib.add_book(Book("Advanced Python", "B", 150))
lib.add_book(Book("Java Fundamentals", "C", 200))

assert lib.count_books_with("python") == 2
assert lib.count_books_with("Java") == 1
````

## Optional Exercise 3: Use a Private Attribute for Rating

**Task**: Modify the `Book` class to include a **private attribute** `__rating` (between 0 and 5). Then:

1. Add a method `set_rating(r)` to set the rating. Raise a `ValueError` if `r` is not between 0 and 5.
2. Add a method `get_rating()` to return the current rating.

This teaches basic encapsulation and data protection using private variables.

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Prefix the attribute with double underscore: <code>self.__rating</code>. Use <code>set_rating()</code> and <code>get_rating()</code> to access it.
</details>

````python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        self.__rating = None

    def set_rating(self, r):
        if r < 0 or r > 5:
            raise ValueError("Rating must be between 0 and 5.")
        self.__rating = r

    def get_rating(self):
        return self.__rating
````

````python
# Tests
b = Book("Encapsulated", "Author", 100)
b.set_rating(4.5)
assert b.get_rating() == 4.5

try:
    b.set_rating(6)
    assert False
except ValueError:
    pass
````
