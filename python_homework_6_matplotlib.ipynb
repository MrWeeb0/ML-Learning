## Exercise 1: Basic Line Plot

**Task**: Write a function `plot_basic_line()` that:

1. Plots the numbers from 0 to 9 on the x-axis.
2. Plots their squares on the y-axis.
3. Returns the `matplotlib.figure.Figure` object.

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>plt.plot(x, y)</code> and <code>plt.figure()</code> to create the plot. Don't forget to return the figure!
</details>

```python
import matplotlib.pyplot as plt

def plot_basic_line():
    fig = plt.figure()
    x = [i for i in range(10)]
    y = [i**2 for i in range(10)]
    plt.plot(x, y)
    return fig
```

```python
# Tests
fig = plot_basic_line()
assert fig is not None
assert isinstance(fig, plt.Figure)
```

## Exercise 2: Add Labels and Title

**Task**: Write a function `plot_labeled_line()` that:

- Plots `x = [0, 1, 2, 3]` and `y = [0, 1, 4, 9]`
- Sets the x-axis label to `'X Axis'`
- Sets the y-axis label to `'Y Axis'`
- Sets the plot title to `'Squared Numbers'`
- Returns the `Figure` object

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>plt.xlabel()</code>, <code>plt.ylabel()</code>, and <code>plt.title()</code> after plotting.
</details>

```python
def plot_labeled_line():
    fig = plt.figure()
    x = [0, 1, 2, 3]
    y = [0, 1, 4, 9]
    plt.plot(x, y)
    plt.xlabel("X Axis")
    plt.ylabel("Y Axis")
    plt.title("Squared Numbers")
    return fig
```

```python
# Tests
fig = plot_labeled_line()
ax = fig.axes[0]
assert ax.get_xlabel() == "X Axis"
assert ax.get_ylabel() == "Y Axis"
assert ax.get_title() == "Squared Numbers"
```

## Exercise 3: Plot Multiple Series

**Task**: Write a function `plot_multiple_series()` that:

- Plots two lines on the same axes:
  - `x = [0, 1, 2, 3, 4]`
  - `y1 = x` (line 1)
  - `y2 = [i**2 for i in x]` (line 2)
- Uses different colors and line styles for each line
- Adds a legend labeling them as `'Linear'` and `'Squared'`
- Returns the `Figure` object

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>label=</code> inside <code>plt.plot()</code> and then <code>plt.legend()</code> to display it.
</details>

```python
def plot_multiple_series():
    fig = plt.figure()
    x = [0, 1, 2, 3, 4]
    y1 = x
    y2 = [i**2 for i in x]
    plt.plot(x, y1, label='Linear', color='blue', linestyle='-')
    plt.plot(x, y2, label='Squared', color='red', linestyle='--')
    plt.legend()
    return fig
```

```python
# Tests
fig = plot_multiple_series()
ax = fig.axes[0]
labels = [text.get_text() for text in ax.get_legend().get_texts()]
assert "Linear" in labels and "Squared" in labels
assert len(ax.lines) == 2
```

## Exercise 4: Scatter Plot with Color

**Task**: Write a function `plot_colored_scatter()` that:

## Exercise 6: Histogram of a Distribution

**Task**: Write a function `plot_histogram()` that:

- Generates 1000 random values from a normal distribution (mean 0, std 1)
- Plots a histogram with 20 bins
- Adds the title `'Normal Distribution'`
- Returns the `Figure` object

<details>
  <summary><strong>Click here for a hint!</strong></summary>

## Exercise 7: Customize Ticks and Grid

**Task**: Write a function `plot_with_grid_and_ticks()` that:

- Plots `x = [0, 1, 2, 3, 4]` and `y = [0, 1, 4, 9, 16]`
- Sets the x-ticks to `[0, 2, 4]`
- Enables a grid on the plot
- Returns the `Figure` object

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>plt.xticks()</code> and <code>plt.grid(True)</code>.
</details>

```python
def plot_with_grid_and_ticks():
    fig = plt.figure()
    x = [0, 1, 2, 3, 4]
    y = [0, 1, 4, 9, 16]
    plt.plot(x, y)
    plt.xticks([0, 2, 4])
    plt.grid(True)
    return fig
```

```python
# Tests
fig = plot_with_grid_and_ticks()
ax = fig.axes[0]

# Check number of custom ticks
assert ax.xaxis.get_ticklocs().shape[0] <= 3
```

## Exercise 8: Save Plot to File

**Task**: Write a function `save_plot_to_file()` that:

- Plots a simple line graph of `x = [0, 1, 2]`, `y = [0, 1, 4]`
- Saves the figure as `'my_plot.png'`
- Returns `True` if the file exists after saving

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>Use <code>plt.savefig()</code>.
</details>

```python
import os

def save_plot_to_file():
    fig = plt.figure()
    x = [0, 1, 2]
    y = [0, 1, 4]
    plt.plot(x, y)
    plt.savefig('my_plot.png')
    plt.close(fig)
    return os.path.exists('my_plot.png')
```

```python
# Tests
assert save_plot_to_file() is True
```

## Exercise 9: Subplots in a Grid

**Task**: Write a function `plot_with_subplots()` that:

- Creates a 1-row, 2-column subplot layout
- The left plot should be a line: `x = [0, 1, 2]`, `y = [0, 1, 4]`
- The right plot should be a bar chart of `'A', 'B', 'C'` with heights `[5, 3, 7]`
- Returns the `Figure` object

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>plt.subplot(1, 2, 1)</code> and <code>plt.subplot(1, 2, 2)</code> to switch between subplots.
</details>

```python
def plot_with_subplots():
    fig = plt.figure()
    
    plt.subplot(1, 2, 1)
    x = [0, 1, 2]
    y = [0, 1, 4]
    plt.plot(x, y)
    
    plt.subplot(1, 2, 2)
    categories = ['A', 'B', 'C']
    values = [5, 3, 7]
    plt.bar(categories, values)
    
    return fig
```

```python
# Tests
fig = plot_with_subplots()
assert isinstance(fig, plt.Figure)
assert len(fig.axes) == 2
```

## Exercise 10: Use Styles and Formatting

**Task**: Write a function `styled_plot()` that:

- Uses the `'ggplot'` style
- Plots `x = [1, 2, 3, 4]` and `y = [10, 20, 15, 25]`
- Sets the marker to `'o'` and the line to dashed (`'--'`)
- Returns the `Figure` object

<details>

  Use <code>plt.style.use('ggplot')</code> and set <code>linestyle='--'</code>, <code>marker='o'</code>.

## Optional Exercise 1: Annotate a Line Plot

**Task**: Write a function `plot_with_annotation()` that:

- Plots the line: `x = np.linspace(0, 10, 100)`, `y = np.sin(x)`
- Adds an annotation on the peak of the sine curve near `x = π/2`
- Adds the text: `"Peak"` with an arrow pointing to the curve
- Returns the `Figure` object

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>plt.annotate()</code> with the <code>arrowprops</code> argument. The peak occurs near <code>x = np.pi/2</code>.
</details>

```python
def plot_with_annotation():
    fig = plt.figure()
    x = np.linspace(0, 10, 100)
    y = np.sin(x)
    plt.plot(x, y)
    
    peak_x = np.pi / 2
    peak_y = np.sin(peak_x)
    plt.annotate('Peak', xy=(peak_x, peak_y), xytext=(peak_x + 1, peak_y + 0.3),
                arrowprops=dict(arrowstyle='->', color='red'))
    return fig
```

```python
# Tests
fig = plot_with_annotation()
ax = fig.axes[0]
assert len(ax.lines) == 1
assert len(ax.texts) >= 1
```

## Optional Exercise 2: Plot Multiple Lines with Legend

**Task**: Write a function `plot_multiple_lines()` that:

- Plots two lines:
  - `y1 = np.sin(x)`,
  - `y2 = np.cos(x)`
  - where `x = np.linspace(0, 2*np.pi, 100)`
- Uses different colors and line styles
- Adds a legend with labels `'Sine'` and `'Cosine'`
- Returns the `Figure` object

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>plt.plot(..., label='Sine')</code> and then call <code>plt.legend()</code>.
</details>

```python
def plot_multiple_lines():
    fig = plt.figure()
    x = np.linspace(0, 2*np.pi, 100)
    y1 = np.sin(x)
    y2 = np.cos(x)
    plt.plot(x, y1, label='Sine', color='blue', linestyle='-')
    plt.plot(x, y2, label='Cosine', color='red', linestyle='--')
    plt.legend()
    return fig
```

```python
# Tests
fig = plot_multiple_lines()
ax = fig.axes[0]
assert len(ax.lines) == 2
```

## Optional Exercise 3: Plot from a DataFrame

**Task**: Write a function `plot_from_dataframe()` that:

- Creates a DataFrame with columns `'Year'`, `'Sales'`, and `'Profit'`
- Years: [2020, 2021, 2022, 2023]
- Sales: [100, 120, 140, 160]
- Profit: [20, 25, 30, 40]
- Plots `'Sales'` and `'Profit'` over `'Year'` on the same figure
- Adds a title `'Company Performance'`
- Returns the `Figure` object

<details>
  <summary><strong>Click here for a hint!</strong></summary>
  Use <code>df.plot()</code> with <code>x='Year'</code> and call <code>plt.title()</code> afterward.
</details>

```python
import pandas as pd

def plot_from_dataframe():
    fig = plt.figure()
    df = pd.DataFrame({
        'Year': [2020, 2021, 2022, 2023],
        'Sales': [100, 120, 140, 160],
        'Profit': [20, 25, 30, 40]
    })
    df.plot(x='Year', y=['Sales', 'Profit'], ax=fig.add_subplot(111))
    plt.title('Company Performance')
    return fig
```

```python
# Tests
fig = plot_from_dataframe()
ax = fig.axes[0]
assert ax.get_title() == "Company Performance"
assert len(ax.lines) == 2
```
