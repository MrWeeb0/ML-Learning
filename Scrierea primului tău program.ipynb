### Scrierea comentariilor în Python


<p>
    În plus față de scrierea codului, este întotdeauna o idee bună să adaugi comentarii în codul tău. Comentariile îi ajută pe alții să înțeleagă ce ai încercat să realizezi (motivul pentru care ai scris un anumit fragment de cod). Acest lucru nu doar că îi ajută pe <strong>ceilalți</strong> să îți înțeleagă codul, ci îți poate servi și ție drept reminder atunci când revii asupra lui, după săptămâni sau chiar luni.
</p>

<p>
    Pentru a scrie comentarii în Python, utilizează simbolul diez <code>#</code> înaintea comentariului propriu-zis. Atunci când rulezi codul, Python va ignora tot ce urmează după simbolul <code>#</code> pe acea linie.
</p>


````python
# Exersează în scrierea comentariilor

print('Hello, Python!') # Această linie afișează un șir de caractere
# print("Hi")
````

<p>
    După executarea celulei de mai sus, vei observa că textul <code>This line prints a string</code> nu a apărut în rezultat, deoarece acesta a fost un comentariu (și prin urmare a fost ignorat de către Python).
</p>
<p>
    Nici a doua linie nu a fost executată, deoarece instrucțiunea <code>print('Hi')</code> a fost precedată, la rândul ei, de simbolul diez (<code>#</code>). Întrucât aici nu este vorba de un comentariu explicativ din partea programatorului, ci de o linie de cod propriu-zisă, putem spune că programatorul a <em>comentat</em> (sau „a pus în comentariu”) acea a doua linie de cod.
</p>


<h3 id="errors">Erori în Python</h3>


<p>Toată lumea face greșeli. Pentru multe tipuri de greșeli, Python te va anunța că ai făcut o eroare prin afișarea unui mesaj specific. Este important să citești cu atenție aceste mesaje de eroare pentru a înțelege exact unde ai greșit și cum poți corecta problema.</p>

<p>De exemplu, dacă tastezi greșit instrucțiunea <code>print</code>, scriind <code>frint</code>, Python va afișa un mesaj de eroare. Încearcă acest exemplu:</p>


````python
# Afișează un mesaj de eroare sub formă de șir de caractere

frint("Hello, Python!")
````

<p>Mesajul de eroare îți indică următoarele informații:</p>
<ol>
    <li>locul în care s-a produs eroarea (acest lucru devine și mai util în celule mari ale notebook-ului sau în scripturi mai lungi);</li>
    <li>tipul erorii apărute (<code>NameError</code>).</li>
</ol>
<p>În exemplul nostru, Python a încercat să ruleze funcția <code>frint</code>, însă nu a putut identifica ce reprezintă <code>frint</code>, deoarece nu este o funcție încorporată și nici nu a fost definită anterior de către noi.</p>


<p>
    Vei observa că, dacă facem un alt tip de greșeală, cum ar fi să uităm să închidem un șir de caractere (string), vom obține un alt tip de eroare (mai exact, o eroare de tip <code>SyntaxError</code>). Încearcă mai jos:
</p>


````python
# Încearcă să vezi un mesaj de eroare încorporat

print("Hello, Python!)
````

### Python știe despre eroarea ta înainte de a rula codul?

Python este ceea ce numim un <em>limbaj interpretat</em>. Limbajele compilate analizează întregul program în timpul compilării și pot avertiza în legătură cu o întreagă clasă de erori înainte de execuție. În contrast, Python interpretează scriptul linie cu linie, pe măsură ce îl execută. Python va opri execuția întregului program în momentul în care întâlnește o eroare (cu excepția cazului în care eroarea este anticipată și gestionată de către programator — un subiect mai avansat pe care îl vom aborda mai târziu în cadrul acestui curs).


Încearcă să rulezi codul din celula de mai jos și vezi ce se întâmplă:

````python
# Afișează un șir și apoi provoacă o eroare pentru a observa ordinea execuției

print("This will be printed")
frint("This will cause an error")
print("This will NOT be printed")
````
